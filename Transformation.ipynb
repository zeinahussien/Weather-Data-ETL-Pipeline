### 2. Transformation

```python
import pandas as pd

df = pd.read_csv("raw_cities_data.csv")

df
```

```python
df['Temperature'] = (df['Temperature'] - 273.15).round(2)
df['Latitude'] = df['Latitude'].round(2)
df['Longitude'] = df['Longitude'].round(2)
```

```python
def region_creation(city):
    if city == 'Cairo':
        return 'Greater Cairo'
    elif city == 'Giza':
        return 'Greater Cairo'
    elif city == 'Alexandria':
        return 'North Coast'
    elif city == 'Port Said':
        return 'Canal'
    elif city == 'Suez':
        return 'Canal'
    elif city == 'Ismailia':
        return 'Canal'
    elif city == 'Mansoura':
        return 'Lower Egypt'
    elif city == 'Tanta':
        return 'Lower Egypt'
    elif city == 'Luxor':
        return 'Upper Egypt'
    elif city == 'Aswan':
        return 'Upper Egypt'

df['Region'] = df['City'].apply(region_creation)
```

```python
def temp_category_creation(temp):
    if temp < 27:
        return 'Cold'
    elif temp < 31:
        return 'Moderate'
    else:
        return 'Hot'

df['TemperatureCategory'] = df['Temperature'].apply(temp_category_creation)
```

```python
def humidity_category_creation(humidity):
    if humidity < 30:
        return 'Low'
    elif humidity < 40:
        return 'Medium'
    else:
        return 'High'

df['HumidityCategory'] = df['Humidity'].apply(humidity_category_creation)

df
```

```python
import datetime

df['IngestionTime'] = datetime.datetime.now()
```

```python
df.to_csv('transformed_cities_data.csv', index=False)
```
