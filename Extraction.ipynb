### 1. Extraction

```python
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("API_TOKEN")
```

```python
list_cities = []
```

```python
cities = [
    "Cairo",
    "Giza",
    "Alexandria",
    "Port Said",
    "Suez",
    "Ismailia",
    "Mansoura",
    "Tanta",
    "Luxor",
    "Aswan"
]
```

```python
import requests

url = "https://api.openweathermap.org/data/2.5/weather"

for i in cities:
    params = {
        "appid": API_KEY,
        "q": i
        }

    response = requests.get(url, params=params)

    data = response.json()

    City = i
    Country = data['sys']['country']
    Latitude = data['coord']['lat']
    Longitude = data['coord']['lon']
    Temperature = data['main']['temp']
    Pressure = data['main']['pressure']
    Humidity = data['main']['humidity']
    WindSpeed = data['wind']['speed']
    WeatherMain = data['weather'][0]['main']
    WeatherDescription = data['weather'][0]['description']

    list_cities.append([
        City,
        Country,
        Latitude,
        Longitude,
        Temperature,
        Pressure,
        Humidity,
        WindSpeed,
        WeatherMain,
        WeatherDescription
        ])

list_cities

```

```python
import pandas as pd

df = pd.DataFrame(list_cities, columns=[
        'City',
        'Country',
        'Latitude',
        'Longitude',
        'Temperature',
        'Pressure',
        'Humidity',
        'WindSpeed',
        'WeatherMain',
        'WeatherDescription'])

df.to_csv("raw_cities_data.csv", index=False)
```
