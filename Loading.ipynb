### 3. Loading

```python
import os
from dotenv import load_dotenv

load_dotenv()

db_server = os.getenv("DB_SERVER")
db_name = os.getenv("DB_NAME")
db_username = os.getenv("DB_USER")
db_password = os.getenv("DB_PASSWORD")
```

```python
import pyodbc
conn = pyodbc.connect(                    
    "driver={ODBC Driver 17 for SQL Server};"
    f"server={db_server};"
    f"database={db_name};"
    f"UID={db_username};"
    f"PWD={db_password};"
)
 
cursor = conn.cursor()
```
