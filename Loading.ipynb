### 3. Loading

```python
import os
from dotenv import load_dotenv

load_dotenv()

db_server = os.getenv("DB_SERVER")
db_name = os.getenv("DB_NAME")
db_username = os.getenv("DB_USER")
db_password = os.getenv("DB_PASSWORD")
```

```python
import pandas as pd
df = pd.read_csv('transformed_cities_data.csv')
rows = list(df.itertuples(index = False, name = None))

rows
```

```python
import pyodbc
conn = pyodbc.connect(                    
    "driver={ODBC Driver 17 for SQL Server};"
    f"server={db_server};"
    f"database={db_name};"
    f"UID={db_username};"
    f"PWD={db_password};"
)
 
cursor = conn.cursor()

cursor.execute("DROP TABLE IF EXISTS transformed_data;")

cursor.execute("""CREATE TABLE transformed_data (
    city varchar(50),
    country varchar(50),
    latitude varchar(50),
    longitude varchar(50),
    temperature varchar(50),
    pressure varchar(50),
    humidity varchar(50),
    wind_speed varchar(50),
    weather_main varchar(50),
    weather_desc varchar(50),
    region varchar(50),
    temperature_category varchar(50),
    humidity_category varchar(50),
    ingestion_time varchar(50)
    )
    
    """)

insert_query = insert_query = """
    INSERT INTO transformed_data 
    VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
"""

cursor.executemany(insert_query, rows)
conn.commit()

cursor.close()
conn.close()

print("All records successfully loaded into the SQL Server.")
```
